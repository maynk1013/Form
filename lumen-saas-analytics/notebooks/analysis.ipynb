```python
import sqlite3
from pathlib import Path

import pandas as pd

ROOT = Path("..").resolve()
conn = sqlite3.connect(ROOT / "data" / "lumen.db")
print((ROOT / "outputs" / "FINDINGS.md").read_text().split("## What the numbers say")[0])
```

```python
bridge = pd.read_sql_query(
    (ROOT / "sql" / "analysis" / "01_mrr_bridge.sql").read_text(),
    conn,
)
gap = (
    bridge["beginning_mrr"] + bridge["new_mrr"] + bridge["reactivation_mrr"]
    + bridge["expansion_mrr"] - bridge["contraction_mrr"] - bridge["churned_mrr"]
    - bridge["ending_mrr"]
).abs().max()
print("Max bridge gap:", gap)
bridge.tail(6)
```
